### Simple LLM call

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from getpass import getpass

os.environ["GOOGLE_API_KEY"] = getpass("Enter your Gemini API key: ")

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",

)

response = llm.invoke(
    "What is an LLM? Answer in exactly 10 words."
)

print(response.text)
```

### Tool calling

```python
response = llm.invoke(
    "What is the current USD to INR exchange rate right now ?."
)

print(response.text)
#the model has not real time data so it say the NO
```

#API call


```python
import requests

rate = requests.get(
    "https://open.er-api.com/v6/latest/usd"
).json()

current_rate = rate["rates"]["INR"]
print(current_rate)
```

```python
response = llm.invoke(
    f"""
    Current USD to INR exchange rate is {current_rate}.
    Explain this excahnge rate in one simple sentence
    """
)
print(response.text)
```
